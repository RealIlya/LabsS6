```python
import math
import matplotlib.pyplot as plt

def target_function(x):
  #return (x - 2)**2
  x_ = x/2.
  return 0.5 - (x_) * math.exp(-(x_**2))

START_A = 0.
START_B = 5.
VARIANT_EPS = 1e-3

def method_dichotomy(func, a, b, eps, verbose=False):
    delta = eps / 2.1  
    calls = 0
    history = []
    iteration = 0
    prev_len = b - a
    
    while (b - a) > eps:
        iteration += 1
        x1 = (a + b - delta) / 2
        x2 = (a + b + delta) / 2
        
        fx1 = func(x1)
        fx2 = func(x2)
        calls += 2
        
        current_len = b - a
        ratio = prev_len / current_len if iteration > 1 else 0
        
        if verbose:
            history.append({
                'i': iteration,
                'x1': x1, 'x2': x2,
                'fx1': fx1, 'fx2': fx2,
                'a': a, 'b': b,
                'len': current_len,
                'ratio': ratio
            })
            
        prev_len = current_len

        if fx1 < fx2:
            b = x2
        else:
            a = x1
            
    min_x = (a + b) / 2
    return min_x, func(min_x), calls, history


def method_golden_section(func, a, b, eps, verbose=False):
    phi = (1 + math.sqrt(5)) / 2 # Золотая пропорция
    calls = 0
    history = []
    iteration = 0
    
    # Инициализация точек
    x1 = b - (b - a) / phi
    x2 = a + (b - a) / phi
    fx1 = func(x1)
    fx2 = func(x2)
    calls += 2
    
    prev_len = b - a
    
    while (b - a) > eps:
        iteration += 1
        current_len = b - a
        ratio = prev_len / current_len if iteration > 1 else 0
        
        if verbose:
            history.append({
                'i': iteration,
                'x1': x1, 'x2': x2,
                'fx1': fx1, 'fx2': fx2,
                'a': a, 'b': b,
                'len': current_len,
                'ratio': ratio
            })
            
        prev_len = current_len

        if fx1 < fx2:
            b = x2
            x2 = x1
            fx2 = fx1
            x1 = b - (b - a) / phi
            fx1 = func(x1)
            calls += 1
        else:
            a = x1
            x1 = x2
            fx1 = fx2
            x2 = a + (b - a) / phi
            fx2 = func(x2)
            calls += 1
            
    min_x = (a + b) / 2
    return min_x, func(min_x), calls, history

def method_fibonacci(func, a, b, eps, verbose=False):
    fibs = [1, 1]
    limit = (b - a) / eps
    while fibs[-1] < limit:
        fibs.append(fibs[-1] + fibs[-2])
    
    n = len(fibs) - 1 
    
    x1 = a + (fibs[n-2] / fibs[n]) * (b - a)
    x2 = a + (fibs[n-1] / fibs[n]) * (b - a)
    
    fx1 = func(x1)
    fx2 = func(x2)
    calls = 2
    
    history = []
    
    for k in range(1, n - 1):
        current_len = b - a
        
        if verbose:
            history.append({
                'i': k,
                'x1': x1, 'x2': x2,
                'fx1': fx1, 'fx2': fx2,
                'a': a, 'b': b,
                'len': current_len,
                'ratio': fibs[n-k+1]/fibs[n-k] if k < n else 0 # Примерное соотношение
            })

        if fx1 > fx2:
            a = x1
            x1 = x2
            fx1 = fx2
            x2 = a + (fibs[n - k - 1] / fibs[n - k]) * (b - a)
            fx2 = func(x2)
            calls += 1
        else:
            b = x2
            x2 = x1
            fx2 = fx1
            x1 = a + (fibs[n - k - 2] / fibs[n - k]) * (b - a)
            fx1 = func(x1)
            calls += 1
            
    delta = eps / 10
    x2 = x1 + delta
    fx2 = func(x2)
    calls += 1
    
    if fx1 > fx2:
        a = x1
    else:
        b = x2
        
    min_x = (a + b) / 2
    
    if verbose:
        history.append({
            'i': n - 1,
            'x1': x1, 'x2': x2,
            'fx1': fx1, 'fx2': fx2,
            'a': a, 'b': b,
            'len': b - a,
            'ratio': 1.0
        })
        
    return min_x, func(min_x), calls, history

def search_interval(func, x0, delta=0.1):
    history = []
    
    f0 = func(x0)
    
    if func(x0 + delta) < f0:
        x1 = x0 + delta
        h = delta
    elif func(x0 - delta) < f0:
        x1 = x0 - delta
        h = -delta
    else:
        return [x0 - delta, x0 + delta], history

    k = 1
    xk = x1
    xk_prev = x0
    history.append({'k': 0, 'x': x0, 'f': f0})
    history.append({'k': 1, 'x': x1, 'f': func(x1)})

    while True:
        h *= 2 
        xk_next = xk + h
        fk_next = func(xk_next)
        fk = func(xk)
        
        history.append({'k': k + 1, 'x': xk_next, 'f': fk_next})
        
        if not (fk < fk_next):
            interval = sorted([xk_prev, xk_next])
            return interval, history
        
        xk_prev = xk
        xk = xk_next
        k += 1

def print_table(history, method_name):
    print(f"\nТаблица результатов: {method_name}")
    print("-" * 105)
    print(f"{'i':<4} | {'x1':^12} | {'x2':^12} | {'f(x1)':^12} | {'f(x2)':^12} | {'a':^12} | {'b':^12} | {'b-a':^12} | {'ratio':^8}")
    print("-" * 105)
    for row in history:
        # ratio может быть 0 на первой итерации
        r_str = f"{row['ratio']:.4f}" if row['ratio'] != 0 else "-"
        print(f"{row['i']:<4} | {row['x1']: .6f} | {row['x2']: .6f} | {row['fx1']: .6f} | {row['fx2']: .6f} | {row['a']: .6f} | {row['b']: .6f} | {row['len']: .6e} | {r_str:^8}")
    print("-" * 105)

def print_interval_table(history):
    print(f"\nТаблица поиска интервала:")
    print("-" * 40)
    print(f"{'step':<5} | {'x':^15} | {'f(x)':^15}")
    print("-" * 40)
    for row in history:
        print(f"{row['k']:<5} | {row['x']: .6f} | {row['f']: .6f}")
    print("-" * 40)

if __name__ == "__main__":
    # Дихотомия
    min_d, val_d, calls_d, hist_d = method_dichotomy(target_function, START_A, START_B, VARIANT_EPS, verbose=True)
    print_table(hist_d, "Метод Дихотомии")
    print(f"Результат: x_min = {min_d}, f(x_min) = {val_d}, Вызовов функции: {calls_d}")

    # Золотое сечение
    min_g, val_g, calls_g, hist_g = method_golden_section(target_function, START_A, START_B, VARIANT_EPS, verbose=True)
    print_table(hist_g, "Метод Золотого Сечения")
    print(f"Результат: x_min = {min_g}, f(x_min) = {val_g}, Вызовов функции: {calls_g}")

    # --- ЗАДАНИЕ 2: График зависимости вычислений от точности ---
    print("\n>>> ПОСТРОЕНИЕ ГРАФИКА ЗАВИСИМОСТИ")
    
    epsilons = [10**(-i) for i in range(1, 8)] # от 10^-1 до 10^-7
    log_eps = [-math.log10(e) for e in epsilons] # для оси X (1, 2, ... 7)
    
    dich_calls_list = []
    gold_calls_list = []
    
    print(f"{'Eps':<10} | {'Dichotomy':<10} | {'Golden':<10}")
```

```python

    for eps in epsilons:
        _, _, c_d, _ = method_dichotomy(target_function, START_A, START_B, eps)
        _, _, c_g, _ = method_golden_section(target_function, START_A, START_B, eps)
        dich_calls_list.append(c_d)
        gold_calls_list.append(c_g)
        print(f"{eps:<10} | {c_d:<10} | {c_g:<10}")

    plt.figure(figsize=(10, 6))
    plt.plot(log_eps, dich_calls_list, marker='o', label='Дихотомия')
    plt.plot(log_eps, gold_calls_list, marker='s', label='Золотое сечение')
    plt.title('Зависимость кол-ва вычислений функции от точности')
    plt.xlabel('-lg(ε)')
    plt.ylabel('Количество вычислений f(x)')
    plt.grid(True)
    plt.legend()
    plt.show() # Раскомментируйте, чтобы показать график в окне
```

```python
import math
import matplotlib.pyplot as plt

def target_function(x):
    return (x - 2)**2

START_A = -2.0
START_B = 20.0
VARIANT_EPS = 1e-7

def method_dichotomy(func, a, b, eps):
    delta = eps / 2.1
    calls = 0
    history = []
    iteration = 0
    prev_len = b - a
    while (b - a) > eps:
        iteration += 1
        current_len = b - a
        ratio = prev_len / current_len if iteration > 1 else 0
        x1 = (a + b - delta) / 2
        x2 = (a + b + delta) / 2
        
        fx1 = func(x1)
        fx2 = func(x2)
        calls += 2  # Дихотомия: всегда 2 вызова за шаг
        
        history.append({
            'i': iteration,
            'x1': x1, 'x2': x2,
            'fx1': fx1, 'fx2': fx2,
            'a': a, 'b': b,
            'len': b - a,
            'ratio': ratio
        })
        prev_len = current_len
        if fx1 < fx2:
            b = x2
        else:
            a = x1
            
    min_x = (a + b) / 2
    return min_x, func(min_x), calls, history


def method_golden_section(func, a, b, eps):
    phi = (1 + math.sqrt(5)) / 2
    history = []
    iteration = 0
    
    # Инициализация (2 вызова)
    x1 = b - (b - a) / phi
    x2 = a + (b - a) / phi
    fx1 = func(x1)
    fx2 = func(x2)
    calls = 2
    prev_len = b - a
    while (b - a) > eps:
        iteration += 1
        current_len = b - a
        ratio = prev_len / current_len if iteration > 1 else 0        
        history.append({
            'i': iteration,
            'x1': x1, 'x2': x2,
            'fx1': fx1, 'fx2': fx2,
            'a': a, 'b': b,
            'len': b - a,
            'ratio': ratio
        })
        prev_len = current_len
        # Переход к следующему шагу (1 вызов)
        if fx1 < fx2:
            b = x2
            x2 = x1
            fx2 = fx1
            x1 = b - (b - a) / phi
            fx1 = func(x1)
            calls += 1
        else:
            a = x1
            x1 = x2
            fx1 = fx2
            x2 = a + (b - a) / phi
            fx2 = func(x2)
            calls += 1
            
    min_x = (a + b) / 2
    return min_x, func(min_x), calls, history


def method_fibonacci(func, a, b, eps):
    fibs = [1, 1]
    limit = (b - a) / eps
    while fibs[-1] < limit:
        fibs.append(fibs[-1] + fibs[-2])
    n = len(fibs) - 1
    
    # Начальные точки (2 вызова)
    x1 = a + (fibs[n-2] / fibs[n]) * (b - a)
    x2 = a + (fibs[n-1] / fibs[n]) * (b - a)
    fx1 = func(x1)
    fx2 = func(x2)
    calls = 2
    
    history = []
    
    # Цикл идет n-2 раза
    for k in range(1, n - 1):
        history.append({
            'i': k,
            'x1': x1, 'x2': x2,
            'fx1': fx1, 'fx2': fx2,
            'a': a, 'b': b,
            'len': b - a,
            'ratio': fibs[n-k+1]/fibs[n-k] if k < n else 0 # Примерное соотношение
        })

        if fx1 > fx2:
            a = x1
            x1 = x2
            fx1 = fx2
            x2 = a + (fibs[n - k - 1] / fibs[n - k]) * (b - a)
            fx2 = func(x2)
            calls += 1
        else:
            b = x2
            x2 = x1
            fx2 = fx1
            x1 = a + (fibs[n - k - 2] / fibs[n - k]) * (b - a)
            fx1 = func(x1)
            calls += 1
            
    delta = eps / 100
    x2 = x1 + delta
    fx2 = func(x2)
    calls += 1
    
    history.append({
        'i': n - 1,
        'x1': x1, 'x2': x2,
        'fx1': fx1, 'fx2': fx2,
        'a': a, 'b': b,
        'len': b - a,
        'ratio': 1.0
    })

    if fx1 > fx2:
        a = x1
    else:
        b = x2
        
    min_x = (a + b) / 2
    return min_x, func(min_x), calls, history

def print_table(history, method_name, total_calls):
  print(f"\nТаблица результатов: {method_name}")
  print(f"Всего вычислений функции: {total_calls}")
  if method_name == "Метод Дихотомии":
    print(f"Проверка: {len(history)} строк * 2 = {len(history)*2} вызовов")
  else:
    print(f"Проверка: 2 (начальных) + {len(history)-1} (в цикле) + {total_calls - (2 + len(history)-1)} (финал) = {total_calls} вызовов")  
  print("-" * 105)
  print(f"{'i':<4} | {'x1':^12} | {'x2':^12} | {'f(x1)':^12} | {'f(x2)':^12} | {'a':^12} | {'b':^12} | {'b-a':^12} | {'ratio':^8}")
  print("-" * 105)
  for row in history:
      # ratio может быть 0 на первой итерации
      r_str = f"{row['ratio']:.4f}" if row['ratio'] != 0 else "-"
      print(f"{row['i']:<4} | {row['x1']: .6f} | {row['x2']: .6f} | {row['fx1']: .6f} | {row['fx2']: .6f} | {row['a']: .6f} | {row['b']: .6f} | {row['len']: .6e} | {r_str:^8}")
  print("-" * 105)

if __name__ == "__main__":
    # Дихотомия
    min_d, val_d, calls_d, hist_d = method_dichotomy(target_function, START_A, START_B, VARIANT_EPS)
    print_table(hist_d, "Метод Дихотомии", calls_d)

    # Золотое сечение
    min_g, val_g, calls_g, hist_g = method_golden_section(target_function, START_A, START_B, VARIANT_EPS)
    print_table(hist_g, "Метод Золотого Сечения", calls_g)

    # Фибоначчи
    min_f, val_f, calls_f, hist_f = method_fibonacci(target_function, START_A, START_B, VARIANT_EPS)
    print_table(hist_f, "Метод Фибоначчи", calls_f)
```

```python
    print("\n>>> СБОР ДАННЫХ ДЛЯ ГРАФИКА")
    epsilons = [10**(-i) for i in range(1, 8)]
    log_eps = [-math.log10(e) for e in epsilons]
    
    dich_calls = []
    gold_calls = []
    fib_calls = []
    
    print(f"{'Eps':<10} | {'Dich':<6} | {'Gold':<6} | {'Fib':<6}")
    for eps in epsilons:
        dich_calls.append(method_dichotomy(target_function, START_A, START_B, eps)[2])
        gold_calls.append(method_golden_section(target_function, START_A, START_B, eps)[2])
        fib_calls.append(method_fibonacci(target_function, START_A, START_B, eps)[2])
        print(f"{eps:<10} | {dich_calls[-1]:<6} | {gold_calls[-1]:<6} | {fib_calls[-1]:<6}")

    plt.figure(figsize=(10, 6))
    plt.plot(log_eps, dich_calls, 'o-', label='Дихотомия')
    plt.plot(log_eps, gold_calls, 's-', label='Золотое сечение')
    plt.plot(log_eps, fib_calls, '^-', label='Фибоначчи')
    plt.xlabel('-lg(ε)')
    plt.ylabel('Вычисления функции')
    plt.legend()
    plt.grid(True)
    print("График построен.")
```
